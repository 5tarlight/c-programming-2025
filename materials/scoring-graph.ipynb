$$ \text{Score} = 1500 \cdot \text{shortest path length} \cdot \left( \frac{1}{\text{time}} + \frac{1}{\text{player move}} \right) $$

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

shortest = 100

taken = np.linspace(1, 200, 100)
move_count = np.linspace(1, 300, 100)

T, M = np.meshgrid(taken, move_count)

time_score = 1500.0 * (shortest / T)
move_score = 1500.0 * (shortest / M)
score = time_score + move_score

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(M, T, score)

ax.set_title("Maze Game Score Surface")
ax.set_xlabel("Move Count")
ax.set_ylabel("Time Taken (s)")
ax.set_zlabel("Score")

# 축 범위 조절
ax.set_xlim(0, 300)
ax.set_ylim(0, 200)
ax.set_zlim(0, 10000)

# 보기 좋게 시점 조절
ax.view_init(elev=30, azim=135)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

shortest = 100

taken = np.linspace(1, 200, 300)
move_count = np.linspace(1, 300, 300)

T, M = np.meshgrid(taken, move_count)

time_score = 1500.0 * (shortest / T)
move_score = 1500.0 * (shortest / M)
score = time_score + move_score

# 점수 범위를 0~10000으로 제한
score = np.clip(score, 0, 10000)

plt.figure(figsize=(10, 6))
contour = plt.contourf(M, T, score, levels=100, cmap='viridis')
cbar = plt.colorbar(contour)
cbar.set_label("Score (clipped to max 10,000)")

plt.title("Maze Game Score Contour")
plt.xlabel("Move Count")
plt.ylabel("Time Taken (s)")
plt.tight_layout()
plt.show()
```
